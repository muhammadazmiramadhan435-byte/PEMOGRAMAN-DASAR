```python
# Latihan 4 - Pengolahan List Tegangan
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Akses elemen pertama & terakhir
print("Data awal     :", tegangan_harian[0], "V")
print("Data akhir    :", tegangan_harian[-1], "V")

# (b) Tambah data baru
tegangan_harian.append(222.0)

# (c) Hitung total elemen
print("Total sampel  :", len(tegangan_harian))
```

```python
# Latihan 5 - Dictionary Spesifikasi Motor
motor_listrik = {
    "nama": "Motor Servo DC",
    "daya_kW": 7.5,
    "tegangan_V": 220,
    "rpm": 1200
}

# Ambil nilai berdasarkan key
print("Kapasitas daya :", motor_listrik["daya_kW"], "kW")

# Update nilai rpm
motor_listrik["rpm"] = 1800

# Tampilkan seluruh dictionary terbaru
print("Data Motor Terbaru :", motor_listrik)
```

```python
# Latihan 6 - List of Dictionary Pembangkit
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"}
]

# 1. Tampilkan nama unit indeks ke-1
print("Unit Kedua:", pembangkit[1]["nama"])

# 2. Ubah status Unit-3
pembangkit[2]["status"] = "standby"

# 3. Tambah unit baru dengan append()
pembangkit.append({"nama": "Unit-Cadangan", "kapasitas_MW": 200, "status": "operasi"})

# 4. Tampilkan semua data
print("Daftar Pembangkit:", pembangkit)
```
