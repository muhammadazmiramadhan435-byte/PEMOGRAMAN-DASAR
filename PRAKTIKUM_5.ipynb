```python
# Program untuk mengecek bonus dan diskon
total_belanja = input("Total belanja: Rp ")
total_belanja = int(total_belanja)

# Jika belanja lebih dari Rp100.000
if total_belanja > 100000:
    print("Selamat! Anda mendapatkan voucher makan.")
    print("Anda mendapatkan diskon 5%.")

    # Menghitung diskon
    diskon = total_belanja * 5 / 100

    # Menghitung total pembayaran
    bayar = total_belanja - diskon
else:
    bayar = total_belanja

# Mencetak struk
print("Total yang harus dibayar: Rp", int(bayar))
print("Terima kasih sudah berbelanja di Toko Kami")
print("Kami tunggu kunjungan Anda selanjutnya.")
```

```python
hari = "Sabtu"

if hari == "Sabtu" or hari == "Minggu":
    diskon = 0.20
    print(f"Selamat! Anda mendapat diskon akhir pekan sebesar {diskon*100}%.")
elif hari == "Jumat":
    diskon = 0.15
    print(f"Diskon spesial Jumat! Anda mendapat potongan {diskon*100}%.")
else:
    diskon = 0.05
    print(f"Anda mendapat diskon hari biasa sebesar {diskon*100}%.")
```

```python
rupiah = int(input("Masukkan jumlah uang (Rp): "))

print("1. Dollar (USD)")
print("2. Euro (EUR)")
print("3. Yen (JPY)")

pilih = int(input("Pilih konversi (1/2/3): "))

if pilih == 1:
    print("Dalam Dollar:", rupiah / 15000)
elif pilih == 2:
    print("Dalam Euro:", rupiah / 16000)
elif pilih == 3:
    print("Dalam Yen:", rupiah / 120)
else:
    print("Pilihan tidak valid")
```
