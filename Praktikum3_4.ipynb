```python

# Input lima nilai mahasiswa
tugas = float(input("Masukkan nilai tugas: "))
kuis = float(input("Masukkan nilai kuis: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))
praktik = float(input("Masukkan nilai praktik: "))

# Menghitung nilai akhir
nilai_akhir = (
    (0.2 * tugas) +
    (0.1 * kuis) +
    (0.2 * uts) +
    (0.3 * uas) +
    (0.2 * praktik)
)

# Menampilkan hasil
print("\n=== Hasil Nilai Mahasiswa ===")
print(f"Nilai Tugas: {tugas}")
print(f"Nilai Kuis: {kuis}")
print(f"Nilai UTS: {uts}")
print(f"Nilai UAS: {uas}")
print(f"Nilai Praktik: {praktik}")
print(f"Nilai Akhir: {nilai_akhir:.2f}")
```

```python

# Program Belanja Buku

jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_buku = int(input("Masukkan harga per buku: "))
uang_azmi = int(input("Masukkan uang azmi: "))

# Menghitung total belanja
total = jumlah_buku * harga_buku

# Menghitung diskon 10%
diskon = total * 10 // 100

# Menghitung jumlah yang harus dibayar
bayar = total - diskon

# Menghitung uang kembalian
kembalian = uang_azmi - bayar

# Menghitung pembagian uang kepada 4 orang
per_orang = bayar // 4
sisa = bayar % 4

# Memeriksa apakah uang Budi cukup
uang_cukup = uang_azmi >= bayar

# Menampilkan hasil
print("\nTotal      :", total)
print("Diskon     :", diskon)
print("Bayar      :", bayar)
print("Kembalian  :", kembalian)
print("Per orang  :", per_orang)
print("Sisa       :", sisa)
print("Uang cukup?:", uang_cukup)
```

```python

# Program Perhitungan PLTS Atap Rumah

# Input data
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya setiap panel (Wp): "))
jam_matahari = float(input("Masukkan jam penyinaran efektif per hari: "))
efisiensi = float(input("Masukkan efisiensi sistem (%): ")) / 100
kebutuhan_harian = float(input("Masukkan kebutuhan listrik per hari (kWh): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))

# Menghitung energi yang dihasilkan per hari
energi_harian = (jumlah_panel * daya_panel * jam_matahari * efisiensi) / 1000

# Menghitung energi per bulan
energi_bulanan = energi_harian * 30

# Menghitung nilai rupiah listrik per bulan
nilai_listrik = energi_bulanan * tarif_listrik

# Menghitung kekurangan energi per hari
kekurangan = kebutuhan_harian - energi_harian

# Memeriksa apakah PLTS mencukupi kebutuhan
mencukupi = energi_harian >= kebutuhan_harian

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN PLTS ===")
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(max(0, kekurangan), 1))
print("Mencukupi?          :", mencukupi)
```

```python

# Program Generator Diesel dan Rugi Kabel

# Memasukkan data
daya = float(input("Masukkan daya generator (kW): "))
jam = float(input("Masukkan lama operasi per hari (jam): "))
konsumsi = float(input("Masukkan konsumsi solar (liter/kWh): "))
harga = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (Ohm): "))

# 1. Menghitung energi per hari
energi = daya * jam

# 2. Menghitung kebutuhan solar dan biaya
solar = energi * konsumsi
biaya = solar * harga

# 3. Menghitung lama tangki mencukupi dan sisa solar
tangki_cukup = kapasitas // solar
sisa_solar = kapasitas % solar

# 4. Menghitung rugi daya kabel
rugi_kabel = arus ** 2 * hambatan

# 5. Memeriksa apakah rugi kabel tergolong tinggi
rugi_tinggi = rugi_kabel > 150 and arus >= 20

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN ===")
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", tangki_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
