```python

# MONTHLY PRODUCTION REPORT (PORTFOLIO)

# Data produksi energi selama 30 hari (MWh)
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

days = list(range(1, 31))

# 1. Total dan rata-rata energi
total = sum(energy)
average = total / len(energy)

print("===== MONTHLY PRODUCTION REPORT =====")
print("Total energy   :", f"{total:,}", "MWh")
print("Average energy :", f"{average:,.2f}", "MWh/day")

# 2. Tiga produksi tertinggi dan terendah
data_sorted = sorted(
    [(energy[i], days[i]) for i in range(len(energy))],
    reverse=True
)

print("\nThree Highest Production:")
for value, day in data_sorted[:3]:
    print(f"Day {day}: {value:,} MWh")

print("\nThree Lowest Production:")
for value, day in sorted(data_sorted, reverse=False)[:3]:
    print(f"Day {day}: {value:,} MWh")

# 3. Memasukkan target energi
target = float(input("\nEnter energy target (MWh): "))

below_target = []

for i in range(len(energy)):
    if energy[i] < target:
        below_target.append(days[i])

print("\nDays below target:", below_target)
print("Number of days below target:", len(below_target))

# 4. Menghitung 7-day moving average
avg_7 = []
below_97 = []

limit = average * 0.97

for i in range(6, len(energy)):
    window = energy[i-6:i+1]
    moving_average = sum(window) / 7
    avg_7.append(moving_average)

    if moving_average < limit:
        below_97.append((i - 5, i + 1, moving_average))

print("\n===== 7-DAY MOVING AVERAGE =====")
print("97% of monthly average:", round(limit, 2), "MWh/day")

print("Days whose moving average is below 97%:")

for start, end, value in below_97:
    print(
        f"Day {start}-{end}: "
        f"{value:,.2f} MWh/day"
    )

# 5. Total energi per minggu
print("\n===== WEEKLY ENERGY =====")

week1 = sum(energy[0:7])
week2 = sum(energy[7:14])
week3 = sum(energy[14:21])
week4 = sum(energy[21:28])
remaining = sum(energy[28:30])

print("Week 1 :", f"{week1:,}", "MWh")
print("Week 2 :", f"{week2:,}", "MWh")
print("Week 3 :", f"{week3:,}", "MWh")
print("Week 4 :", f"{week4:,}", "MWh")
print("Remaining days :", f"{remaining:,}", "MWh")

print("\n===== REPORT COMPLETED =====")
```
